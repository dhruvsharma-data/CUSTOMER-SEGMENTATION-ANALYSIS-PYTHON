```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Mall_Customers.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
print("Total Customers:", len(df))

print("Average Spending:", df['Spending Score (1-100)'].mean())
```

```python
high = df[df['Spending Score (1-100)'] > 70]
print("High spenders:", len(high))
```

```python
df['Age Group'] = pd.cut(df['Age'],
                        bins=[18,25,35,50,70],
                        labels=['18-25','26-35','36-50','50+'])

df['Age Group'].value_counts()
```

```python
import matplotlib.pyplot as plt

plt.hist(df['Spending Score (1-100)'])
plt.title("Spending Score Distribution")
plt.show()
```

```python
plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'])
plt.title("Income vs Spending")
plt.show()
```

Customer Segmentation Analysis

Introduction

This project focuses on analyzing customer data to understand spending behavior and identify high-value customer segments using Python.

Dataset Description

The dataset contains information about customers such as Gender, Age, Annual Income, and Spending Score. It helps in understanding how different factors influence customer purchasing behavior.

Data Analysis

- Performed data loading and basic exploration using Pandas
- Checked for missing values and data types
- Analyzed customer distribution and spending patterns
- Identified high spending customers
- Created age groups to understand behavior across different segments

Visualization

- Used histogram to analyze distribution of spending scores
- Used scatter plot to study relationship between income and spending

Key Insights

- High spending customers form an important segment
- Customers aged 26–35 tend to spend more
- Income and spending are not directly proportional
- Some low-income customers also show high spending behavior

Conclusion

This analysis helps in understanding customer behavior and can be used by businesses to target high-value customers and improve marketing strategies.
